```python
# Electricity Bill Calculator

def calculate_bill(customer_name, units, rate_per_unit, fixed_charge=50):
    """
    Calculate electricity bill for a customer.
    :param customer_name: str - Name of the customer
    :param units: float - Units consumed (kWh)
    :param rate_per_unit: float - Rate per unit (₹)
    :param fixed_charge: float - Fixed charge applied to all bills
    :return: dict - Bill details
    """
    # Validation
    if units < 0 or rate_per_unit < 0:
        raise ValueError("Units and rate must be non-negative.")

    # Calculation
    bill_amount = (units * rate_per_unit) + fixed_charge

    # Prepare bill details
    return {
        "Customer": customer_name,
        "Units Consumed": units,
        "Rate per Unit": rate_per_unit,
        "Fixed Charge": fixed_charge,
        "Total Bill": bill_amount
    }


# Example usage
customer_name = "Anushree"
units = 250   # kWh
rate_per_unit = 6.5  # ₹ per kWh

bill = calculate_bill(customer_name, units, rate_per_unit)

# Print bill
print("\n--- Electricity Bill ---")
for key, value in bill.items():
    print(f"{key}: {value}")
print("------------------------")
```
